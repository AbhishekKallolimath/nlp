```python
----lab 1----
import os
import re

folder = "input_files"
roll = "20241COM1661"

with open("output.txt", "w") as out:
    out.write(roll + "\n")

    for i in range(1, 5):
        filename = "File" + str(i) + ".txt"
        path = os.path.join(folder, filename)

        with open(path, "r", encoding="utf-8") as file:
            lines = file.readlines()

        line_count = len(lines)
        text = ""

        for line in lines:
            if line.strip() != "":
                text = text + line.strip() + " "

        word_count = len(text.split())
        sentence_count = len(re.findall(r"[.?]", text))

        out.write(filename + ":\n")
        out.write("Lines = " + str(line_count) + "\n")
        out.write("Words = " + str(word_count) + "\n")
        out.write("Sentences = " + str(sentence_count) + "\n\n")

print("Done")

```

```python
---lab 2----

import os
import nltk
import spacy

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

nlp = spacy.load("en_core_web_sm")

folder = "input_files"

for i in range(1, 5):

    filename = "File" + str(i) + ".txt"
    path = os.path.join(folder, filename)

    with open(path, "r", encoding="utf-8") as file:
        text = file.read()

    sentences = nltk.sent_tokenize(text)
    words = nltk.word_tokenize(text)
    tags = nltk.pos_tag(words)

    nouns = 0
    verbs = 0
    adjectives = 0
    adverbs = 0

    for word, tag in tags:

        if tag.startswith("NN"):
            nouns += 1
        elif tag.startswith("VB"):
            verbs += 1
        elif tag.startswith("JJ"):
            adjectives += 1
        elif tag.startswith("RB"):
            adverbs += 1

    doc = nlp(text)

    spacy_nouns = 0
    spacy_verbs = 0
    spacy_adjectives = 0
    spacy_adverbs = 0

    for token in doc:

        if token.pos_ == "NOUN" or token.pos_ == "PROPN":
            spacy_nouns += 1
        elif token.pos_ == "VERB":
            spacy_verbs += 1
        elif token.pos_ == "ADJ":
            spacy_adjectives += 1
        elif token.pos_ == "ADV":
            spacy_adverbs += 1

    print("\nFile:", filename)

    print("\nNLTK")
    print("Sentences =", len(sentences))
    print("Words =", len(words))
    print("Nouns =", nouns)
    print("Verbs =", verbs)
    print("Adjectives =", adjectives)
    print("Adverbs =", adverbs)

    print("\nspaCy")
    print("Nouns =", spacy_nouns)
    print("Verbs =", spacy_verbs)
    print("Adjectives =", spacy_adjectives)
    print("Adverbs =", spacy_adverbs)

```

```python
--- lab 3 ----

import nltk

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tag import pos_tag

nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("averaged_perceptron_tagger_eng")

text = "The rain in Spain stays mainly in the plain. What a glorious feeling of singing in the rain. I am singing and dancing in the rain."

sentences = sent_tokenize(text)
words = word_tokenize(text)

print("Sentences:")
print(sentences)

print("\nWords:")
print(words)

text_lower = text.lower()
words_lower = word_tokenize(text_lower)

print("\nLowercase:")
print(words_lower)

stop_words = stopwords.words("english")
words_without_stopwords = []

for word in words_lower:
    if word not in stop_words:
        words_without_stopwords.append(word)

print("\nWithout stopwords:")
print(words_without_stopwords)

print("\nWith stopwords =", len(words))
print("Without stopwords =", len(words_without_stopwords))

stemmer = PorterStemmer()
stemmed_words = []

for word in words:
    stemmed_words.append(stemmer.stem(word))

print("\nStemmed words:")
print(stemmed_words)

lemmatizer = WordNetLemmatizer()
lemmatized_words = []

for word in words:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print("\nLemmatized words:")
print(lemmatized_words)

def get_pos(tag):

    if tag.startswith("J"):
        return "a"

    if tag.startswith("V"):
        return "v"

    if tag.startswith("R"):
        return "r"

    return "n"

tagged_words = pos_tag(words)
lemmatized_text = []

for word, tag in tagged_words:
    new_word = lemmatizer.lemmatize(word, pos=get_pos(tag))
    lemmatized_text.append(new_word)

print("\nPOS Lemmatized words:")
print(lemmatized_text)

```

```python
---lab 4 ---

import gensim.downloader
import numpy as np

model = gensim.downloader.load("glove-twitter-100")

def get_vector(word):

    try:
        return model[word]
    except KeyError:
        print("Word not found")
        return None

def most_similar(word):

    try:
        return model.most_similar(word)[0][0]
    except KeyError:
        print("Word not found")
        return None

def similarity(word1, word2):

    try:
        return model.similarity(word1, word2)
    except KeyError:
        print("Word not found")
        return None

def cosine_similarity(word1, word2):

    v1 = get_vector(word1)
    v2 = get_vector(word2)

    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

def analogy(a, b, c):

    return model.most_similar(
        positive=[b, c],
        negative=[a]
    )[0][0]

print("Vector:")
print(get_vector("king"))

print("\nMost similar word:")
print(most_similar("king"))

print("\nSimilarity:")
print(similarity("king", "queen"))
print(similarity("king", "man"))

print("\nCosine similarity:")
print(cosine_similarity("king", "queen"))

print("\nAnalogy:")
print(analogy("man", "king", "woman"))


```
