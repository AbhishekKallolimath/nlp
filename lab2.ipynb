```python
import os
import nltk
import spacy
from collections import Counter
```

```python
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
```

```python
nlp = spacy.load("en_core_web_sm")
folder = "input_files"
```

```python
def process_file(filename):

    path = os.path.join(folder, filename)

    with open(path, "r", encoding="utf-8") as file:
        text = file.read()

    sentences = nltk.sent_tokenize(text)
    words = nltk.word_tokenize(text)

    tags = nltk.pos_tag(words)

    nltk_counts = Counter()

    for word, tag in tags:

        if tag.startswith("NN"):
            nltk_counts["Nouns"] += 1

        elif tag.startswith("VB"):
            nltk_counts["Verbs"] += 1

        elif tag.startswith("JJ"):
            nltk_counts["Adjectives"] += 1

        elif tag.startswith("RB"):
            nltk_counts["Adverbs"] += 1

    doc = nlp(text)

    spacy_counts = Counter()

    for token in doc:

        if token.pos_ == "NOUN" or token.pos_ == "PROPN":
            spacy_counts["Nouns"] += 1

        elif token.pos_ == "VERB":
            spacy_counts["Verbs"] += 1

        elif token.pos_ == "ADJ":
            spacy_counts["Adjectives"] += 1

        elif token.pos_ == "ADV":
            spacy_counts["Adverbs"] += 1

    print("\nFile:", filename)

    print("\nNLTK")
    print("Sentences:", len(sentences))
    print("Words:", len(words))
    print("Nouns:", nltk_counts["Nouns"])
    print("Verbs:", nltk_counts["Verbs"])
    print("Adjectives:", nltk_counts["Adjectives"])
    print("Adverbs:", nltk_counts["Adverbs"])

    print("\nspaCy")
    print("Nouns:", spacy_counts["Nouns"])
    print("Verbs:", spacy_counts["Verbs"])
    print("Adjectives:", spacy_counts["Adjectives"])
    print("Adverbs:", spacy_counts["Adverbs"])
```

```python
for i in range(1, 5):
    process_file("File" + str(i) + ".txt")
```
