```python
import gensim.downloader
import numpy as np
```

```python
model = gensim.downloader.load("glove-twitter-100")

word1 = "king"
word2 = "queen"
word3 = "man"
```

```python
def getWordVector(word):

    try:
        return model[word]
    except KeyError:
        print("Key error:", word, "is not there")
        return None
```

```python
print("Word Vector for King =", getWordVector(word1))
```

```python
def getMostSimilarWord(word):

    try:
        return model.most_similar(word)[0][0]
    except KeyError:
        print("Key error:", word, "is not there")
        return None
```

```python
print("Most similar word for King is", getMostSimilarWord(word1))
```

```python
def getSimilarity(word1, word2):

    try:
        return model.similarity(word1, word2)
    except KeyError:
        print("Key error: Either", word1, "or", word2, "is not there")
        return None
```

```python
def getVectorCosineSimilarity(word1, word2):

    try:
        wv1 = getWordVector(word1)
        wv2 = getWordVector(word2)

        dp = np.dot(wv1, wv2)
        l1 = np.linalg.norm(wv1)
        l2 = np.linalg.norm(wv2)

        return dp / (l1 * l2)

    except:
        return None
```

```python
print("Similarity between", word1, "and", word2, "is", getSimilarity(word1, word2))
print("Similarity between", word1, "and", word3, "is", getSimilarity(word1, word3))

print("Cosine Similarity between", word1, "and", word2, "is", getVectorCosineSimilarity(word1, word2))
print("Cosine Similarity between", word1, "and", word3, "is", getVectorCosineSimilarity(word1, word3))
```

```python
def completeAnalogy(wordA, wordB, wordC):

    return model.most_similar(
        positive=[wordC, wordB],
        negative=[wordA]
    )[0][0]
```

```python
print("Answer is", completeAnalogy(word1, word2, word3))
```

```python
print("Answer is", completeAnalogy(word1, word2, word3))
```

```python
def getWordVector(word):

    try:
        return model[word]
    except KeyError:
        print("Key error:", word, "is not there")
        return None
```

```python
def getVectorSimilarity(v1, v2):

    return np.dot(v1, v2) / (
        np.linalg.norm(v1) * np.linalg.norm(v2)
    )
```

```python
def completeAnalogyWithOptions(wordA, wordB, wordC, options):

    wvA = getWordVector(wordA)
    wvB = getWordVector(wordB)
    wvC = getWordVector(wordC)

    answerVector = wvC + wvB - wvA

    bestOption = options[0]

    maxSim = getVectorSimilarity(
        getWordVector(bestOption),
        answerVector
    )

    for option in options:

        sim = getVectorSimilarity(
            getWordVector(option),
            answerVector
        )

        if sim > maxSim:
            maxSim = sim
            bestOption = option

    return bestOption
```

```python
print(completeAnalogyWithOptions(
    "pantry",
    "store",
    "scullery",
    ["kitchen", "cook", "utensils", "wash"]
))
```
