```python
---Lab 1 — File Handling-------------
import os
import re

folder = "input_files"
output_file = "output.txt"
roll_number = "20241COM1661"

def process_file(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        lines = file.readlines()

    line_count = len(lines)
    flattened = ""

    for line in lines:
        if line.strip() != "":
            flattened += line.strip() + " "

    words = flattened.split()
    word_count = len(words)
    sentence_count = len(re.findall(r"[.?]", flattened))

    return line_count, word_count, sentence_count

with open(output_file, "w", encoding="utf-8") as output:
    output.write(roll_number + "\n")

    for i in range(1, 5):
        filename = f"File{i}.txt"
        file_path = os.path.join(folder, filename)

        lines, words, sentences = process_file(file_path)

        output.write(f"{filename}:\n")
        output.write(f"Lines = {lines}\n")
        output.write(f"Words = {words}\n")
        output.write(f"Sentences = {sentences}\n")
        output.write("\n")

print(f"Output generated successfully: {output_file}")
---Lab 2 — NLTK and spaCy----------------
import os
import nltk
import spacy
from collections import Counter

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

nlp = spacy.load("en_core_web_sm")

input_folder = "input_files"

def process_file(filename):

    filepath = os.path.join(input_folder, filename)

    with open(filepath, "r", encoding="utf-8") as file:
        text = file.read()

    sentences = nltk.sent_tokenize(text)
    sentence_count = len(sentences)

    words = nltk.word_tokenize(text)
    word_count = len(words)

    nltk_pos_tags = nltk.pos_tag(words)

    nltk_counts = Counter()
    nltk_counts["Nouns"] = 0
    nltk_counts["Adjectives"] = 0
    nltk_counts["Verbs"] = 0
    nltk_counts["Adverbs"] = 0

    for sentence in sentences:
        words = nltk.word_tokenize(sentence)
        taggedSentence = nltk.pos_tag(words)

        for word, tag in taggedSentence:
            if tag.startswith("NN"):
                nltk_counts["Nouns"] += 1
            elif tag.startswith("VB"):
                nltk_counts["Verbs"] += 1
            elif tag.startswith("JJ"):
                nltk_counts["Adjectives"] += 1
            elif tag.startswith("RB"):
                nltk_counts["Adverbs"] += 1

    doc = nlp(text)

    spacy_counts = Counter()

    for token in doc:

        if token.pos_ == "NOUN":
            spacy_counts["Nouns"] += 1
        elif token.pos_ == "PROPN":
            spacy_counts["Nouns"] += 1
        elif token.pos_ == "VERB":
            spacy_counts["Verbs"] += 1
        elif token.pos_ == "ADJ":
            spacy_counts["Adjectives"] += 1
        elif token.pos_ == "ADV":
            spacy_counts["Adverbs"] += 1

    print("=" * 60)
    print("File:", filename)
    print("=" * 60)

    print("\nNLTK Results")
    print("-" * 30)
    print("Number of sentences :", sentence_count)
    print("Number of words     :", word_count)
    print("Number of nouns     :", nltk_counts["Nouns"])
    print("Number of verbs     :", nltk_counts["Verbs"])
    print("Number of adjectives:", nltk_counts["Adjectives"])
    print("Number of adverbs   :", nltk_counts["Adverbs"])

    print("\nspaCy Results")
    print("-" * 30)
    print("Number of nouns     :", spacy_counts["Nouns"])
    print("Number of verbs     :", spacy_counts["Verbs"])
    print("Number of adjectives:", spacy_counts["Adjectives"])
    print("Number of adverbs   :", spacy_counts["Adverbs"])

files = ["File1.txt", "File2.txt", "File3.txt", "File4.txt"]

for filename in files:
    process_file(filename)
---Lab 3 — Corpus Cleaning----------------
text = "The rain in Spain stays mainly in the plain. What a glorious feeling of singing in the rain. I am singing and dancing in the rain."

import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

sentences = sent_tokenize(text)
words = word_tokenize(text)
print(sentences)
print(words)

text_lower = text.lower()
words_lower = word_tokenize(text_lower)
print(words_lower)

from nltk.corpus import stopwords

stop_words = stopwords.words('english')
words_without_stopwords = []

for word in words_lower:
    if word not in stop_words:
        words_without_stopwords.append(word)

print("Number of words with stopwords = ", len(words))
print("Number of words without stopwords = ", len(words_without_stopwords))

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
stemmed_words = []

for word in words:
    stemmed_words.append(stemmer.stem(word))

print(stemmed_words)

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized_words = []

for word in words:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print(lemmatized_words)

def getPoSFromTag(tag):
    if tag.startswith('J'):
        return 'a'
    if tag.startswith('V'):
        return 'v'
    if tag.startswith('R'):
        return 'r'
    return 'n'

from nltk.tag import pos_tag

tagged_words = pos_tag(words)
lemmatized_text = []

for word, tag in tagged_words:
    lemmatized_word = lemmatizer.lemmatize(word, pos=getPoSFromTag(tag))
    lemmatized_text.append(lemmatized_word)

print(lemmatized_text)
--------Lab 4 — Word Vector--------
!pip install nltk
!pip install numpy
!pip install gensim

modelName = "glove-twitter-100"

import gensim.downloader

model = gensim.downloader.load(modelName)

word1 = "king"
word2 = "queen"
word3 = "man"

def getWordVector(word):
    try:
        return model[word]
    except:
        print("Key error ", word, " is not there")

print("Word Vector for King = ", str(getWordVector(word1)))
print("Word Vector for ASDFS = ", str(getWordVector("ASDFS")))

def getMostSimilarWord(word):
    try:
        return model.most_similar(word)[0][0]
    except KeyError:
        print("Key error", word, "is not there")
        return None

print("Most similar word for King is", str(getMostSimilarWord(word1)))
print("Most similar word for ASDFD is", str(getMostSimilarWord("ASDFD")))

import numpy as np

def getSimilarity(word1, word2):
    try:
        return model.similarity(word1, word2)
    except KeyError:
        print("Key error: Either", word1, "or", word2, "is not there")

def getVectorCosineSimilarity(word1, word2):
    try:
        wv1 = getWordVector(word1)
        wv2 = getWordVector(word2)
        dp = np.dot(wv1, wv2)
        l1 = np.linalg.norm(wv1)
        l2 = np.linalg.norm(wv2)
        cs = dp/(l1 * l2)
        return cs
    except KeyError:
        print("Key error: Either", word1, "or", word2, "is not there")
        return None

print("Similarity between", word1, "and", word2, "is", str(getSimilarity(word1, word2)))
print("Similarity between", word1, "and", word3, "is", str(getSimilarity(word1, word3)))
print("Cosine Similarity between", word1, "and", word2, "is", str(getVectorCosineSimilarity(word1, word2)))
print("Cosine Similarity between", word1, "and", word3, "is", str(getVectorCosineSimilarity(word1, word3)))

def completeAnalogy(wordA, wordB, wordC):
    return model.most_similar(positive=[wordC, wordB], negative=[wordA])[0][0]

print("Answer is", str(completeAnalogy(word1, word2, word3)))

import gensim.downloader

modelName = "fasttext-wiki-news-subwords-300"
model = gensim.downloader.load(modelName)

import numpy as np

def getWordVector(word):
    try:
        return model[word]
    except:
        print("Key error ", word, " is not there")

def getVectorSimilarity(v1, v2):
    return np.dot(v1, v2)/(np.linalg.norm(v1) * np.linalg.norm(v2))

def completeAnalogyWithOptions(wordA, wordB, wordC, options):
    wvA = getWordVector(wordA)
    wvB = getWordVector(wordB)
    wvC = getWordVector(wordC)
    answerVector = wvC + wvB - wvA
    bestOption = options[0]
    maxSim = getVectorSimilarity(getWordVector(bestOption), answerVector)

    for option in options:
        sim = getVectorSimilarity(getWordVector(option), answerVector)

        if sim > maxSim:
            maxSim = sim
            bestOption = option

    return bestOption

print(completeAnalogyWithOptions("pantry", "store", "scullery", ["kitchen", "cook", "utensils", "wash"]))
print(completeAnalogyWithOptions("calf", "cow", "puppy", ["dog", "bitch", "donkey", "mare"]))
print(completeAnalogyWithOptions("paw", "cat", "hoof", ["horse", "lion", "lamb", "elephant"]))
print(completeAnalogyWithOptions("carpenter", "saw", "tailor", ["sew", "cloth", "needle", "tape"]))
print(completeAnalogyWithOptions("coal", "heat", "wax", ["candle", "bee", "energy", "light"]))
print(completeAnalogyWithOptions("traveler", "journey", "sailor", ["water", "voyage", "ship", "crew"]))
print(completeAnalogyWithOptions("election", "manifesto", "meeting", ["agenda", "minutes", "circular", "preface"]))
print(completeAnalogyWithOptions("squint", "eye", "squeeze", ["tongue", "cloth", "hand", "throat"]))
print(completeAnalogyWithOptions("antiseptic", "germs", "antidote", ["poison", "wound", "allergy", "infection"]))
```
